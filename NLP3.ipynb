```python

print("saad khokar uin 231a019")
import emoji
text_no_emoji = emoji.replace_emoji(text_num, '')
print(text_no_emoji)
```

```python
import nltk
nltk.download("punkt_tab")
```

```python
from nltk import word_tokenize
import nltk
nltk.download("punkt")
tokenized=word_tokenize(text_no_emoji)
print("tokenized words:",tokenized)
```

```python
print("saad khokar uin 231a019")
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("saad khokar uin 231a019")
filtered_tokens = [word for word in tokenized if word not in punctuation]
print("Filtered tokens (without punctuation):", filtered_tokens)
```

```python
print("saad khokar uin 231a019")
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
print(stopwords.words("english"))
```

```python
print("saad khoar uin 231a019")
text_clean2 = []

for word  in filtered_tokens:
  if word not in stopwords.words("english"):
    text_clean2.append(word)
print(text_clean2)
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
print([stemmer.stem(word) for word in text_clean2])
```
