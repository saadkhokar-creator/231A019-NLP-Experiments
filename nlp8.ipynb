```python
print("Saad Khokar UIN:231A019")
documents=[
"Machine learning enables computers to learn from data.",
"Computers can learn patterns from data using machine learning.",
"Natural language processing helps computers understand human language.",
"The football team won the match yesterday.",
"Deep learning is a branch of machine learning."
]
```

```python
print("Saad Khokar UIN:231A019")
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
print("Saad Khokar UIN:231A019")
for i, document in enumerate(documents):
   print(f"D{i+1}: {document}")
```

```python
print("Saad Khokar UIN:231A019")
vectorizer=TfidfVectorizer(stop_words='english')
tfidf_matrix=vectorizer.fit_transform(documents)
print(tfidf_matrix.shape)
```

```python
print("Saad Khokar UIN:231A019")
tfidf_df=pd.DataFrame(

tfidf_matrix.toarray(),
columns=vectorizer.get_feature_names_out(),
index=["D1", "D2", "D3", "D4", "D5"]
)
tfidf_df.round(2)
```

```python
print("Saad Khokar UIN:231A019")
print(tfidf_matrix)
```

```python
print("Saad Khokar UIN:231A019")
print(vectorizer.get_feature_names_out())
```

```python
print("Saad Khokar UIN:231A019")
similarity_matrix=cosine_similarity(tfidf_matrix)
print(similarity_matrix)
```

```python
print("Saad Khokar UIN:231A019")
similarity_df = pd.DataFrame (
similarity_matrix,
index=["D1", "D2", "D3", "D4", "D5"],
columns=["D1", "D2", "D3", "D4", "D5"]
)
similarity_df.round(3)
```

```python
print("Saad Khokar UIN:231A019")
highest_similarity = -1
most_similar_pair = None
for i in range(len(documents)):
    for j in range(i + 1, len(documents)):
        similarity = similarity_matrix[i][j]
        if similarity > highest_similarity:
            highest_similarity = similarity
            most_similar_pair = (i, j)

print(
    f"Most similar documents:\n"
    f"D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}"
)
print(f"Similarity: {highest_similarity:.3f}")
```

```python
print("Saad Khokar UIN:231A019")
def similar(D1, D2):
    vector = vectorizer.transform([D1, D2])
    similarity = cosine_similarity(vector)
    return similarity[0, 1]

```
