```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
```

```python
import nltk
nltk.download('punkt_tab')
text="Meta is looking to acquire a new  AI startup in germany,Elon Musk was also interested in this same startup"
tokens=word_tokenize (text)
print(tokens)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
tagged_tokens = pos_tag(tokens)
print(tagged_tokens)
```

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
named_entities=ne_chunk(tagged_tokens)
print(named_entities)
```

```python
#Display named entities
for subtree in named_entities:
  if isinstance(subtree, nltk.Tree):
    entity = " ".join([word for word, tag in subtree.leaves()])
    label = subtree.label()
    print(f"Entity: {entity}, Label: {label}")
```

```python
import spacy

nlp=spacy.load("en_core_web_sm")

text="Meta is looking to acquire a new  AI startup in germany,Elon Musk was also interested in this same startup for $100 Billion"


#Process the text
doc=nlp(text)
# Extract named entities
for ent in doc.ents:
  print(f"Entity: {ent.text}, Label: {ent.label_}")
```

```python
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser

sentence="The quick brown fox jumps over the lazy dog."
```

```python
tokens=word_tokenize(sentence)
tagged= pos_tag(tokens)
tagged
```

```python
grammar= "NP: {<DT>?<JJ>*<NN>}"
```

```python

chunk_parser =RegexpParser (grammar)
chunked = chunk_parser.parse(tagged)
# Display the chunked output
print(chunked)
```
